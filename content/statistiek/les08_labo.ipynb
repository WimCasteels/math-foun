Voor deze oefening gebruik je de functie `stats.norm.cdf(x, mu, sigma)` uit scipy. Deze berekent $P(X \leq x)$ voor een normale verdeling met gemiddelde `mu` en standaarddeviatie `sigma`.

Handige patronen:
- $P(X \leq x)$: `stats.norm.cdf(x, mu, sigma)`
- $P(X > x)$: `1 - stats.norm.cdf(x, mu, sigma)`
- $P(a < X < b)$: `stats.norm.cdf(b, mu, sigma) - stats.norm.cdf(a, mu, sigma)`
- De $x$-waarde waarboven een percentage $p$ scoort: `stats.norm.ppf(1 - p, mu, sigma)`
